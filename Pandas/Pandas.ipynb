# Indexing and Selecting Data in Pandas DataFrame

Indexing in pandas means simply selecting particular rows and columns of data from a DataFrame. It allows us to access subsets of data such as:

Selecting all rows and some columns.

Selecting some rows and all columns.

Selecting a specific subset of rows and columns.

Indexing can also be known as Subset Selection.

**1. Indexing a Dataframe using indexing operator []** 

The indexing operator [] is the basic way to select data in Pandas. We can use this operator to access columns from a DataFrame. This method allows us to retrieve one or more columns. The .loc and .iloc indexers also use the indexing operator to make selections.

In order to select a single column, we simply put the name of the column in-between the brackets.

````python
data = pd.read_csv(r'nba.csv' , index_col = "Name")
first = data["Age"]
print(first)
````

The .loc method is used to select data by label. This means it uses the row and column labels to access specific data points. .loc[] is versatile because it can select both rows and columns simultaneously based on labels.

In order to select a single row using .loc[], we put a single row label in a .loc function.

````python
data = pd.read_csv(
    r"nba.csv",
    index_col="Name"
)

first = data.loc["Avery Bradley"]
second = data.loc["R.J. Hunter"]

print(first)
print("\n\n")
print(second)
````

````python
data = pd.read_csv(
    r"nba.csv",
    index_col="Name"
)
row2 = data.iloc[3]
print(row2)
````

# Working with Missing Data

Missing Data can occur when no information is available for one or more items or for an entire row/column. In Pandas missing data is represented as NaN (Not a Number). Missing data can be problematic in real-world datasets where data is incomplete. Pandas provides several methods to handle such missing data effectively:

**1. Checking for Missing Values using isnull() and notnull()**

To check for missing values (NaN) we can use two useful functions:

isnull(): It returns True for NaN (missing) values and False otherwise.
notnull(): It returns the opposite, True for non-missing values and False for NaN values.

Now we drop rows with at least one Nan value (Null value).

````python
import pandas as pd 
import numpy as np

dict = {
    'First Score' : [100 , 90 , np.nan , 95],
    'Second Score' : [30 , np.nan,45 , 56],
    'Third Score ': [52 , 40 , 80 , 98],
    'Fourth Score ' : [np.nan, np.nan, np.nan, 65]
}
df = pd.DataFrame(dict)
print(df)
print("\n")
print(df.dropna())
````

In Pandas, missing data occurs when some values are missing or not collected properly and these missing values are represented as:

None: A Python object used to represent missing values in object-type arrays.

NaN: A special floating-point value from NumPy which is recognized by all systems that use IEEE floating-point standards.

**Checking Missing Values in Pandas**

Pandas provides two important functions which help in detecting whether a value is NaN helpful in making data cleaning and preprocessing easier in a DataFrame or Series are given below :

**1. Using isnull()**
isnull() returns a DataFrame of Boolean value where True represents missing data (NaN). This is simple if we want to find and fill missing data in a dataset.

Example 1: Finding Missing Values in a DataFrame

We will be using Numpy and Pandas libraries for this implementation.

````python
import pandas as pd
import numpy as np
d = {'First Score': [100, 90, np.nan, 95],
        'Second Score': [30, 45, 56, np.nan],
        'Third Score': [np.nan, 40, 80, 98]}
df =pd.DataFrame(d)
mv = df.isnull()
print(mv)
````

````python
import pandas as pd
d= pd.read_csv("employees.csv")
bool_Series = pd.isnull(d["Gender"])
missing_gender_data = d[bool_Series]
print(missing_gender_data)
````

# 3. Using interpolate()

### `interpolate()`

`interpolate()` is a Pandas function used to fill missing (`NaN`) values by estimating them from the surrounding values.

By default, it uses **linear interpolation**, meaning it finds a value between the previous and next known values.

**Example:**
`[10, 20, NaN, 40]` → `[10, 20, 30, 40]`

**Easy to remember:**

> `interpolate()` = **estimate missing values using nearby values.**


````python
import pandas as pd
   
df = pd.DataFrame({"A": [12, 4, 5, None, 1], 
                   "B": [None, 2, 54, 3, None], 
                   "C": [20, 16, None, 3, 8], 
                   "D": [14, 3, None, None, 6]}) 
df.interpolate(method = 'linear' , limit_direction ='forward')
````

# Iterating over rows and columns

Iteration refers to the process of accessing each item one at a time. In Pandas, it means iterating through rows or columns in a DataFrame to access or manipulate the data. We can iterate over rows and columns to extract values or perform operations on each item.

### 1. Iterating Over Rows
There are several ways to iterate over the rows of a Pandas DataFrame and three common methods are:

`iteritems()`
`iterrows()`
`itertuples()`

Each method provides different ways to iterate over the rows which depends on our specific needs.

### 1. Using iterrows()
`iterrows()` returns each row as a `(index, Series)` pair. This means you get the row’s index and a Pandas Series object containing column names and their values. It is easy to use since you can access fields by column name, but it can be slower on very large DataFrames.

Example 1: This code demonstrates how to loop through a small subset of rows and view all column values in each row.

`data.head(2)` selects first 2 rows of the DataFrame.
`for i, row in` .. `iterrows()` gives i as the row index and row as a Series with column-value pairs.

````python
data = pd.read_csv("nba.csv")

# iterate over rows 
for i, row in data.head(2).iterrows():
    print(i,row)
    print()
````

Example 2: This code applies a row-wise condition to filter high-salary players.

loop iterates over every row using iterrows().
pd.notnull(s) ensures missing values (NaN) are skipped.
list of (Name, Salary) is created for players earning above 5M.

````python
high_salary = []
for i , row in data.iterrows():
    s =row['Salary']
    if pd.notnull(s) and s >5_000_000:
        high_salary.append((row['Name'] , int(s)))

for name,sal in high_salary[:10]:
    print(f"{name} | {sal}")
````

### 2. Using `itertuples()`

`itertuples()` returns each row as a named tuple. It is faster than iterrows() because it avoids creating Series objects and you can access row values as attributes (e.g., row.Name, row.Salary). This makes it ideal for performance-heavy tasks.

Example 1: This code shows how each row is returned in tuple format.

Each row is returned as a Pandas namedtuple.
You can access fields directly using attribute syntax (e.g., row.Name).
Much faster than `iterrows()` for larger datasets.

````python
import pandas as pd
data = pd.read_csv("nba.csv")

for row in data.head(3).itertuples():
    print(row)
````

Example 2: This code demonstrates how to extract values from a specific column using list comprehension.

List comprehension iterates through `itertuples()`
.
row.Name extracts the Name field directly.

[:6] limits the result to the first 6 names.

````python
import pandas as pd
data = pd.read_csv("nba.csv")

names = [row.Name for row in data.itertuples()][:6]
print('First six names: \n' , names)
````

### Iterate Over Columns
Iterating columns is useful for column-wise checks or operations (e.g., computing stats, converting dtypes or applying functions). You can iterate columns by using items() or by creating a column-name list.

`1. Using items()`

items() iterates over DataFrame columns, returning each as a (label, Series) pair. The label is column name and Series contains values for that column. This is helpful when you want to check column-level details such as data distribution, types or missing values.

Example 1: This code prints a few values from every column for a quick inspection.

`data.head(2).items()`  loops through each column of the first 2 rows.

label gives the column name, while values is a Series of that column’s data.

````python
import pandas as pd  
data = pd.read_csv("nba.csv") 

for label , values in data.head(2).items():
    print(f"Column: {label}")
    print(values)
    print()
````

Example 2: This example helps detect data quality issues across columns.

`values.dtype` shows the data type of each column.

`values.notnull().sum()` counts how many values are non-missing.

````python
import pandas as pd  
data = pd.read_csv("nba.csv")  

for label , values in data.items(): 
    print(f"{label} : dtype={values.dtype}, non-null = {values.notnull().sum()}")
````


### 2. Using direct column iteration

In Pandas, one can also iterate through columns directly without using methods. This is done by looping over list of column names and then accessing their values from DataFrame.

Example 1: This example prints 3rd row’s value for each column, one column at a time.

`list(data.columns)` gives all column names.
data[col][2] accesses the 3rd row (index 2) value for that column.

````python
import pandas as pd  
data = pd.read_csv("nba.csv") 
cols = list(data.columns)

for col in cols: 
    print(f"{col} : {data[col][2]}")
````

Example 2: This code summarizes numeric columns by calculating their mean values.

select_dtypes(include='number') selects numeric columns.
The loop calculates and prints the mean of each numeric column.

````python
import pandas as pd  
data = pd.read_csv("nba.csv") 

num = data.select_dtypes(include='number')
for col in num.columns:
    print(f"{col} mean:{num[col].mean()}")
````

**Easy way to remember:**

> `loc` → **label**
> `iloc` → **integer position**

### Deprecated `ix[]`

`ix[]` was previously used to select data using either labels or integer positions. It has been **deprecated and removed from modern Pandas**.

Use:

* `loc[]` for label-based selection
* `iloc[]` for position-based selection

Merging DataFrame

Merging DataFrames in Pandas is similar to performing SQL joins. It is useful when we need to combine two DataFrames based on a common column or index. The merge() function provides flexibility for different types of joins.

There are four basic ways to handle the join (inner, left, right and outer) depending on which rows must retain their data.

# Pandas — Merge and Join DataFrames

Combining DataFrames means **bringing related information from two DataFrames into one**.

---

## 1. `merge()` — Combine Using Columns

Use `pd.merge()` when two DataFrames have a **common column/key**.

### One Key

```python
result = pd.merge(df1, df2, on='key')
```

Example:

```text
df1                    df2
key  Name              key  Address
K0   Jai               K0   Nagpur
K1   Princi            K1   Kanpur
```

```python
pd.merge(df1, df2, on='key')
```

Result:

```text
key  Name    Address
K0   Jai     Nagpur
K1   Princi  Kanpur
```

### Multiple Keys

If matching requires more than one column:

```python
result = pd.merge(
    df1,
    df2,
    on=['key', 'key1']
)
```

The row is matched only when **both `key` AND `key1` match**.

---

# 2. `how` in `merge()`

`how` decides **which rows to keep**.

| `how`     | Meaning      | Keeps                     |
| --------- | ------------ | ------------------------- |
| `'inner'` | Intersection | Only matching rows        |
| `'left'`  | Left join    | All left + matching right |
| `'right'` | Right join   | All right + matching left |
| `'outer'` | Union        | All rows from both        |

### Inner

```python
pd.merge(df1, df2, on='key', how='inner')
```

➡️ Only matching keys.

### Left

```python
pd.merge(df1, df2, on='key', how='left')
```

➡️ Keep **everything from left DataFrame**.

### Right

```python
pd.merge(df1, df2, on='key', how='right')
```

➡️ Keep **everything from right DataFrame**.

### Outer

```python
pd.merge(df1, df2, on='key', how='outer')
```

➡️ Keep **everything from both**. Missing values become `NaN`.

### Easy memory trick

```text
inner → matching only
left  → keep left
right → keep right
outer → keep everything
```

---

# 3. `join()` — Combine Using Index

`join()` is mainly used when DataFrames are related through their **index**.

```python
result = df1.join(df2)
```

Example:

```text
df1                 df2
K0  Jai             K0  Nagpur
K1  Princi          K2  Kanpur
K2  Gaurav          K3  Delhi
```

```python
df1.join(df2)
```

Matches using the **index**.

By default, `join()` uses a **left join**.

---

## `join()` with `how`

```python
df1.join(df2, how='outer')
```

Keeps all indexes from both DataFrames.

The same basic options apply:

```text
left
right
inner
outer
```

---

# 4. `join()` with `on`

You can use a column from the left DataFrame to match the **index of the right DataFrame**.

```python
df1.join(df2, on='Key')
```

Think:

```text
df1['Key']  →  df2.index
```

---

# 5. Multi-Index Join

A DataFrame can have more than one index level:

```python
pd.MultiIndex.from_tuples(
    [('K0','Y0'), ('K1','Y1'), ('K2','Y2')]
)
```

Then `join()` can work with these **multi-level indexes**.

You generally don't need to master MultiIndex joins yet.

# Merge vs Join

| Feature       | `merge()`                  | `join()`               |
| ------------- | -------------------------- | ---------------------- |
| Main purpose  | Combine using columns/keys | Combine using indexes  |
| Common syntax | `pd.merge()`               | `df.join()`            |
| Matching      | Columns                    | Index                  |
| Multiple keys | Yes                        | Less commonly used     |
| `how`         | left/right/inner/outer     | left/right/inner/outer |

# How to Sort Pandas DataFrame?

Sorting a DataFrame means arranging its rows in a particular order based on the **values of a column** or the **index**.

Pandas provides two important methods for sorting:

* `sort_values()` → sorts DataFrame by column values
* `sort_index()` → sorts DataFrame by index

---

## 1. Sorting DataFrame Using `sort_values()`

The `sort_values()` method is used to sort a DataFrame based on the values of one or more columns.

### Example

````python
import pandas as pd

data = {
    'Name': ['John', 'Alice', 'Bob', 'David'],
    'Age': [25, 20, 30, 22],
    'Salary': [50000, 60000, 45000, 70000]
}

df = pd.DataFrame(data)

print(df)
````

Now sort the DataFrame according to the `Age` column:


````python
df1 = df.sort_values(by='Age')
df1
````

The rows are arranged from the **smallest Age to the largest Age**.

By default:

```text
ascending = True
```

---

## 2. Sorting in Descending Order

We can use the `ascending` parameter to control the sorting order.

````python
df1 = df.sort_values(by='Age' , ascending= False)
df1
````

Here:

```text
ascending=True   → smallest → largest
ascending=False  → largest → smallest
```

For example, sorting `Salary` from highest to lowest:

````python
df2 = df.sort_values(by = 'Salary' , ascending = False)
df2
````

## 3. Sorting Using Multiple Columns

We can sort a DataFrame using more than one column by passing a **list of column names**.

````python
df3 = df.sort_values(by = [ 'Age' , 'Salary'])
df3
````

Here Pandas first sorts by `Age`.

If two rows have the same `Age`, it then uses `Salary` to decide their order.

We can also specify different sorting orders for each column:

```python
df.sort_values(
    ['Age', 'Salary'],
    ascending=[True, False]
)
```

This means:

```text
Age     → ascending
Salary  → descending
```

---

## 4. Sorting DataFrame by Index

The `sort_index()` method is used to sort the DataFrame according to its **index**.

````python
df1 = df.sort_index(ascending = True)
df1
````

By default, the index is sorted in ascending order.

For descending order:

```python
df.sort_index(ascending=False)
```

---

## 5. Sorting and Updating the Original DataFrame

By default, sorting returns a **new DataFrame** and does not change the original DataFrame.

We can store the result:

````python
df = df.sort_values('Age')
df
````

````python
df.sort_values('Age', inplace=True)
df
````

## 6. Handling Missing Values While Sorting

If a column contains `NaN` values, `sort_values()` allows us to decide where they should appear.

```python
df.sort_values('Salary', na_position='first')
```

Here, missing values are placed at the **beginning**.

We can also use:

```python
df.sort_values('Salary', na_position='last')
```

`last` is the default.

# Binary Operations on Pandas DataFrame and Series

Binary operations involve applying a mathematical or logical operation between **two objects**, such as two Pandas `Series` or two `DataFrame` objects, to produce a new result.

Common binary operations include:

* **Arithmetic operations** — addition, subtraction, multiplication, division, etc.
* **Comparison operations** — equal to, not equal to, greater than, less than, etc.
* **Logical operations** — AND, OR, etc.

Pandas performs these operations **element-wise**, meaning the operation is performed on individual values.

---

# Binary Operations on Pandas Series

## 1. Arithmetic Operations on Series

Arithmetic operations between two Series are performed **element-wise**.

An important feature of Pandas is that it aligns the Series using their **index labels** before performing the operation.

If the indexes do not match, Pandas produces `NaN` for the missing index.

### Example: Adding Two Series

```python
import pandas as pd

s1 = pd.Series([10, 20, 30], index=['a', 'b', 'c'])
s2 = pd.Series([1, 2, 3], index=['a', 'b', 'c'])

result = s1 + s2

print(result)
```

### Output

```text
a    11
b    22
c    33
dtype: int64
```

### Explanation

Pandas matches the values using their **index labels**:

```text
a → 10 + 1 = 11
b → 20 + 2 = 22
c → 30 + 3 = 33
```

So:

```python
s1 + s2
```

performs addition element-by-element.

---

## 2. Series with Different Indexes

Now consider two Series whose indexes are different.

```python
import pandas as pd

s1 = pd.Series([10, 20, 30], index=['a', 'b', 'c'])
s2 = pd.Series([1, 2, 3], index=['a', 'b', 'd'])

result = s1 + s2

print(result)
```

### Output

```text
a    11.0
b    22.0
c     NaN
d     NaN
dtype: float64
```

### Why does this happen?

Pandas aligns the indexes first:

| Index |  s1 |  s2 |
| ----- | --: | --: |
| a     |  10 |   1 |
| b     |  20 |   2 |
| c     |  30 | NaN |
| d     | NaN |   3 |

Therefore:

```text
a → 10 + 1 = 11
b → 20 + 2 = 22
c → 30 + NaN = NaN
d → NaN + 3 = NaN
```

### Important

Pandas uses **index alignment**, not simply the physical position of values.

This is one of the most important ideas to understand about Pandas operations.

### 📌 Pandas DataFrame.`drop_duplicates()`

- 🔹 You can specify columns to check for duplicates.

````python
data = {
    "Name" : ["Alice" , "Bob" , "Alice" , "David"],
    "Age":[25 , 30 , 25 , 40],
    "City" : ["NY" , "LA" , "NY" , "Chicago"]
}
df = pd.DataFrame(data)
print("original dataFrame : ")

print(df)

df_cleaned = df.drop_duplicates()
print("\nModified DataFrame (no duplicates)")
print(df_cleaned)
````

### 📌 Pandas DataFrame.drop_duplicates()

#### 📝 Syntax
DataFrame.drop_duplicates(subset=None, keep='first', inplace=False)

---

### ⚙️ Parameters
1. **subset** → columns to check for duplicates  
   - Default: all columns  
   - Example: `subset=["Name"]`

2. **keep** → which duplicate to retain  
   - `'first'` → keeps first occurrence (default)  
   - `'last'` → keeps last occurrence  
   - `False` → removes all duplicates  

3. **inplace** → whether to modify original DataFrame  
   - `True` → modifies in place  
   - `False` → returns new DataFrame (default)  

---

### 🌟 Return Type
- Returns a **new DataFrame** with duplicates removed (unless `inplace=True`).  

---

### 🔹 Examples
1. **Drop duplicates on all columns**  
   `df.drop_duplicates()`

2. **Drop duplicates based on specific column**  
   `df.drop_duplicates(subset=["Name"])`

3. **Keep last occurrence**  
   `df.drop_duplicates(keep="last")`

4. **Remove all duplicates**  
   `df.drop_duplicates(keep=False)`


````python
df_cleaned = df.drop_duplicates(subset = ["Name"])

print(df_cleaned)
````

#### 🔹 Data Normalization with Pandas
- **Purpose → "Scale for Fairness"**  
  Ensures all numeric features contribute equally in ML models.  

- **Benefits → "Stable, Fast, Equal"**  
  - Avoids numerical instability  
  - Speeds up convergence in gradient algorithms  
  - Equalizes feature contribution  

- **Steps → "Import → Load → Normalize → Visualize"**  
  1. Import libraries (`pandas`, `numpy`, `sklearn.preprocessing`)  
  2. Load or create dataset  
  3. Apply normalization techniques  
  4. Visualize results  






````python
import pandas as pd  
import matplotlib.pyplot as plt  

df = pd.DataFrame([
    [180000, 110, 18.9, 1400],  
    [360000, 905, 23.4, 1800],  
    [230000, 230, 14.0, 1300],  
    [60000, 450, 13.5, 1500]
], columns=['Col A', 'Col B', 'Col C', 'Col D'])

print(df)

df.plot(kind='bar')
plt.show()
````

#### 🔹 Normalization Techniques in Pandas

**1. Maximum Absolute Scaling → "MaxAbs Shrink"**  
- Rescales each feature by dividing values with the **maximum absolute value** in that column.  
- Output range: **–1 to 1**  
- Preserves sparsity (good for sparse data).  




````python
max_scaled = df.copy()

for column in max_scaled.columns:
    max_scaled[column] = max_scaled[column] / max_scaled[column].abs().max()

print(max_scaled)
max_scaled.plot(kind = 'bar')
plt.show()
````

#### 🔹 Normalization Techniques in Pandas

**2. Min‑Max Scaling → "Stretch to [0,1]"**  
- Rescales values to a fixed range **[0,1]**.  
- Formula: `(x – min) / (max – min)`  
- Useful for distance‑based models (KNN, SVM).  


````python
scaled = df.copy()

for column in scaled.columns:
    scaled[column] = (scaled[column] - scaled[column].min()) /( scaled[column].max() - scaled[column].min())

print(scaled)
scaled.plot(kind = 'bar')
plt.show
````

#### 🔹 Normalization Techniques in Pandas

**3. Z‑Score Standardization → "Center & Scale"**  
- Transforms values so that:  
  - Mean = **0**  
  - Standard Deviation = **1**  
- Formula: `(x – mean) / std`  
- Best when data follows a **normal distribution** or when you want to measure how far values are from the average.  




````python
z_scaled = df.copy()
for column in z_scaled.columns:
    z_scaled[column] = (z_scaled[column] - z_scaled[column].mean()) / z_scaled[column].std()

print(z_scaled)
z_scaled.plot(kind = 'bar')
plt.show()
````

  
- **Benefit → "Fair & Fast Models"**  
  - Logistic Regression → coefficients easier to interpret  
  - SVM → distance calculations balanced  
  - Neural Networks → faster convergence, stable gradients  

**Result → "Equal Playing Field"**  
No feature dominates due to scale; all are expressed in terms of how far they are from the average.

#### 🔹 Types of Joins in Pandas

- **Inner Join** → Returns rows with matching values in both DataFrames.  
- **Left Outer Join** → All rows from left DataFrame + matching rows from right.  
- **Right Outer Join** → All rows from right DataFrame + matching rows from left.  
- **Full Outer Join (Outer Join)** → All rows from both DataFrames, matching where possible, NaN where not.  
- **Index Join** → Joins DataFrames based on their index instead of columns.  


````python
#inner Join
a =pd.DataFrame({
    'id' : [1,2,10,12],
    'val1' : ['a' , 'b' , 'c' , 'd']
})

b= pd.DataFrame({
    'id' : [1,2,9,8],
    'val2' : ['p' , 'q' , 'r' , 's']
})
c= pd.merge(a , b , on ='id' , how = 'inner')
print(c)
````

````python
#ledt Join
a =pd.DataFrame({
    'id' : [1,2,10,12],
    'val1' : ['a' , 'b' , 'c' , 'd']
})

b= pd.DataFrame({
    'id' : [1,2,9,8],
    'val2' : ['p' , 'q' , 'r' , 's']
})
c= pd.merge(a , b , on ='id' , how = 'left')
print(c)
````

````python
#right Join
a =pd.DataFrame({
    'id' : [1,2,10,12],
    'val1' : ['a' , 'b' , 'c' , 'd']
})

b= pd.DataFrame({
    'id' : [1,2,9,8],
    'val2' : ['p' , 'q' , 'r' , 's']
})
c= pd.merge(a , b , on ='id' , how = 'right')
print(c)
````

````python
#Outer Join
a =pd.DataFrame({
    'id' : [1,2,10,12],
    'val1' : ['a' , 'b' , 'c' , 'd']
})

b= pd.DataFrame({
    'id' : [1,2,9,8],
    'val2' : ['p' , 'q' , 'r' , 's']
})
c= pd.merge(a , b , on ='id' , how = 'outer')
print(c)
````

#### 🔹 Advanced Operations in Pandas

#### 🔹 Pandas DataFrame.corr()

- **Purpose** → Calculates correlation between numeric columns.  
- **Result** → Returns a **correlation matrix** (DataFrame).  
- **Range** → Values between -1 and +1:  
  - **+1** → perfect positive correlation  
  - **-1** → perfect negative correlation  
  - **0** → no correlation  
- **Non‑numeric columns** → ignored automatically.  
- **Usage** → `df.corr()`  
- **Benefit** → Identifies relationships between variables, useful for analysis & ML feature selection.  


````python
d = {"A": [10, 20, 30], "B": [15, 25, 35]}
df = pd.DataFrame(d)

r = df.corr()
print(r)
````

**Syntax** →  
`DataFrame.corr(method='pearson', min_periods=1, numeric_only=False)`

**Parameters** →  
- **method** → correlation type:  
  - `'pearson'` (default) → linear correlation  
  - `'spearman'` → rank correlation  
  - `'kendall'` → ordinal correlation  
- **min_periods** → minimum required matching values.  
- **numeric_only** → if True, only numeric columns included.  

**Example** →  
- If `Column B = Column A + 5` → perfect linear relation → correlation = **1**.  
- `df[['Height','Weight']].corr()` → shows strong positive correlation. 

````python
import pandas as pd
d = {"Height": [150, 160, 170, 180], "Weight": [50, 65, 68, 80]}
df = pd.DataFrame(d)
print(df.corr())
````

- Correlation = **0.97** → strong positive relation (Height ↑, Weight ↑).  
- Pearson uses **covariance of actual values** → linear correlation.  

#### 🔹 Example 2: Negative Correlation (Kendall)

````python
d = pd.DataFrame({'StudyHours' : [2 ,4 ,6 ,8] , "StressLevel" : [8 , 7 , 5 , 6]})
print(d.corr(method = "kendall"))
````

#### 🔹 Example 3: Spearman Rank Correlation

- **Purpose** → Measures correlation based on **rank order** of values.  
- **Range** → -1 to +1 (same as Pearson/Kendall).  
- **Method** → Non‑parametric, good for **monotonic relationships** (values move in same direction, not necessarily linear).  
- **Usage** →  
  ```python
  df.corr(method='spearman')


````python
import pandas as pd
d = {"MathMarks": [95, 70, 85, 60, 80], "SportsScore": [60, 75, 65, 80, 70]}

df = pd.DataFrame(d)
print(df.corr(method = "spearman"))
````

#### 🔹 Best Practices for Efficient Plotting

   - Customize titles, labels, colors, and gridlines for clarity.  
   - Avoid excessive customization that obscures the data.

4. **Performance Tips**
   - Reduce dataset size (sampling, aggregation) for faster rendering.

   - Always label axes and legends.
